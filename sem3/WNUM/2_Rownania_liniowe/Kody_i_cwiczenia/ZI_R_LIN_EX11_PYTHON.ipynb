### EX11: Metoda Gausa-Seidla
Metoda Gausa-Seidla to iteracyjny algorytm rozwiązywania układów równań liniowych $Ax = b$.
W każdym kroku aktualizuje współrzedną $x_i$ używając **nowych** wartości $x_1,\dots,x_{i-1}$
oraz **starych** wartości $x_{i+1},\dots,x_n$:

$$x_i^{(k+1)} = \frac{1}{a_{ii}}\left(b_i - \sum_{j=1}^{i-1} a_{ij} x_j^{(k+1)} - \sum_{j=i+1}^{n} a_{ij} x_j^{(k)}\right)$$

W przeciwieństwie do metody Jacobi, metoda Gausa-Seidla wykorzystuje najświeższe dostępne
przybliżenia, co zwykle przyspiesza zbieżność. Warunkiem dostatecznym zbieżności jest dominacja wierszowa lub kolumnowa macierzy $A$.

**Macierz Wilkinsona** $W_n$ (ta sama co w EX10): na przekątnej elementy $n, n-1, \dots, 1$, na sąsiednich przekątnych $-1$, reszta zerowa. W ćwiczeniu używamy $A = W_5 + 10I$.
Wektor prawych stron $b = A\mathbf{1}$, czyli rozwiązanie dokładne to $x^* = (1,1,1,1,1)^T$.

```python
import numpy as np
import matplotlib.pyplot as plt

def wilkinson(n):
    A = np.diag(np.arange(n, 0, -1), 0) + np.diag(-np.ones(n-1), 1) + np.diag(-np.ones(n-1), -1)
    return A.astype(np.float64)

A = wilkinson(5) + 10 * np.eye(5)
b = A.sum(axis=1)
x0 = np.random.rand(5)
x = x0.copy()
n = len(x)
delta0 = 1e-8
i_max = 50
err = [np.inf]

for iteration in range(i_max):
    if err[-1] < delta0:
        break
    x_old = x.copy()
    for i in range(n):
        sigma = 0
        for j in range(i):
            sigma += A[i, j] * x[j]       # use new values
        for j in range(i+1, n):
            sigma += A[i, j] * x_old[j]   # use old values
        x[i] = (1/A[i,i]) * (b[i] - sigma)
    err.append(np.linalg.norm(x_old - x))

plt.semilogy(range(1, len(err)), err[1:])
plt.xlabel('# iteracji')
plt.ylabel('Estymata błędu')
plt.grid(True)
plt.show()
```

### ĆWICZENIE 2 — Porównanie z metodą Richardsona

Porównamy szybkość zbieżności metody Gausa-Seidla z metodą Richardsona $x^{(k+1)} = M x^{(k)} + w$, gdzie $M = I - pA$, $w = pb$, dla $p = 0.06$. Na jednym wykresie wyświetlimy estymaty błędu obu metod.

```python
p = 0.06
M = np.eye(5) - p * A
w_vec = p * b
x_old = x0.copy()
err_R = []
i = 1
error = np.inf
while (error >= delta0) and (i < 50):
    x_new = M @ x_old + w_vec
    error = np.linalg.norm(x_new - x_old)
    x_old = x_new
    err_R.append(error)
    i += 1

plt.semilogy(range(1, len(err)), err[1:], label='Gauss-Seidel')
plt.semilogy(range(1, len(err_R)+1), err_R, label=f'Richardson (p={p})')
plt.legend()
plt.xlabel('# iteracji')
plt.ylabel('Estymata błędu')
plt.grid(True)
plt.show()
```
