## EX-6 Porównanie stabilności numerycznej faktorizacji QR i LU przy rozwiązywaniu układów $Ax=b$


```python
import numpy as np
import time
from scipy.linalg import lu_factor, lu_solve

def hilb(n):
    """Generuje macierz Hilberta n x n."""
    return np.array([[1.0/(i+j+1) for j in range(n)] for i in range(n)], dtype=np.float64)

def solveUsingQRFactorization(A, b):
    """Rozwiąż układ Ax = b za pomocą faktorizacji QR."""
    timeStart = time.time()
    Q, R = np.linalg.qr(A)
    timeSimple = time.time() - timeStart
    x = np.linalg.solve(R, Q.T @ b)
    timeAll = time.time() - timeStart
    exact = np.ones(A.shape[0])
    error = np.linalg.norm(x - exact) / np.linalg.norm(exact)
    return timeAll, timeSimple, error

def solveUsingLUFactorization(A, b):
    """Rozwiąż układ Ax = b za pomocą faktorizacji LU."""
    timeStart = time.time()
    lu, piv = lu_factor(A)
    timeSimple = time.time() - timeStart
    x = lu_solve((lu, piv), b)
    timeAll = time.time() - timeStart
    exact = np.ones(A.shape[0])
    error = np.linalg.norm(x - exact) / np.linalg.norm(exact)
    return timeAll, timeSimple, error
```

```python
import numpy as np
import matplotlib.pyplot as plt

Nmax = 200
step = 20

bladQR_1 = []
bladLU_1 = []
bladQR_2 = []
bladLU_2 = []
N = []

for i in range(4, Nmax + 1, step):
    # Przypadek korzystny: losowa dobrze uwarunkowana macierz + 10*I
    # (w wersji MATLAB użyto magic(i) + 10*eye(i); numpy nie ma odpowiednika
    # magic(), więc zastąpiono go macierzą losową — obie ilustrują ten sam
    # przypadek dobrze uwarunkowanej macierzy, ale dają inne liczby na wykresie)
    A1 = np.random.RandomState(i * 137).rand(i, i) + 10 * np.eye(i)
    b1 = np.sum(A1, axis=1)

    # Przypadek niekorzystny: macierz Hilberta
    A2 = hilb(i)
    b2 = np.sum(A2, axis=1)

    t_all, t_simple, err = solveUsingQRFactorization(A1, b1)
    bladQR_1.append(err)

    t_all, t_simple, err = solveUsingLUFactorization(A1, b1)
    bladLU_1.append(err)

    t_all, t_simple, err = solveUsingQRFactorization(A2, b2)
    bladQR_2.append(err)

    t_all, t_simple, err = solveUsingLUFactorization(A2, b2)
    bladLU_2.append(err)

    N.append(i)
    if i % (10 * step) == 0 or i == N[-1]:
        print(f"Postęp: {i}/{Nmax}")
print("Obliczenia zakończone.")
```

```python
import matplotlib.pyplot as plt

Nplt = np.array(N)
errQR1 = np.array(bladQR_1) * 100 + 1e-16
errLU1 = np.array(bladLU_1) * 100 + 1e-16
errQR2 = np.array(bladQR_2) * 100 + 1e-16
errLU2 = np.array(bladLU_2) * 100 + 1e-16

# Figura 1 — skala logarytmiczna
fig1, axes1 = plt.subplots(2, 1, figsize=(8, 8))
axes1[0].semilogy(Nplt, errQR1, "b", label="QR")
axes1[0].semilogy(Nplt, errLU1, "r", label="LU")
axes1[0].set_title("Przypadek korzystny")
axes1[0].set_xlabel("Rozmiar macierzy N")
axes1[0].set_ylabel("Błąd w procentach")
axes1[0].legend()
axes1[0].grid(True)
axes1[1].semilogy(Nplt, errQR2, "b", label="QR")
axes1[1].semilogy(Nplt, errLU2, "r", label="LU")
axes1[1].set_title("Przypadek niekorzystny")
axes1[1].set_xlabel("Rozmiar macierzy N")
axes1[1].set_ylabel("Błąd w procentach")
axes1[1].legend()
axes1[1].grid(True)
fig1.tight_layout()
plt.show()

# Figura 2 — skala liniowa
fig2, axes2 = plt.subplots(2, 1, figsize=(8, 8))
axes2[0].plot(Nplt, errQR1, "b", label="QR")
axes2[0].plot(Nplt, errLU1, "r", label="LU")
axes2[0].set_title("Przypadek korzystny")
axes2[0].set_xlabel("Rozmiar macierzy N")
axes2[0].set_ylabel("Błąd w procentach")
axes2[0].legend()
axes2[0].grid(True)
axes2[1].plot(Nplt, errQR2, "b", label="QR")
axes2[1].plot(Nplt, errLU2, "r", label="LU")
axes2[1].set_title("Przypadek niekorzystny")
axes2[1].set_xlabel("Rozmiar macierzy N")
axes2[1].set_ylabel("Błąd w procentach")
axes2[1].legend()
axes2[1].grid(True)
fig2.tight_layout()
plt.show()
```

## Wykorzystywane funkcje
| Funkcja | Biblioteka | Opis |
|---------|-----------|------|
| `solveUsingQRFactorization(A, b)` | własna | Rozwiązuje układ $Ax=b$ poprzez faktorizację QR |
| `solveUsingLUFactorization(A, b)` | własna | Rozwiązuje układ $Ax=b$ poprzez faktorizację LU |
| `np.linalg.qr(A)` | NumPy | Faktorizacja macierzy $A$ na $Q \cdot R$ |
| `scipy.linalg.lu_factor(A)` | SciPy | Faktorizacja macierzy $A$ na $P \cdot L \cdot U$ |
| `scipy.linalg.lu_solve((lu, piv), b)` | SciPy | Rozwiązuje układ po faktorizacji LU |
| `hilb(n)` | własna | Generuje macierz Hilberta o wymiarze $n \times n$ |
| `np.linalg.norm(v)` | NumPy | Norma euklidesowa wektora |
| `matplotlib.pyplot.subplots(2, 1)` | Matplotlib | Tworzy siatkę 2x1 do rysowania wykresów |
| `axes.semilogy(x, y)` | Matplotlib | Wykres z logarytmiczną osią $y$ |


## Wnioski
1. **Przypadek korzystny** — macierz losowa wzbogacona o $10I$ jest dobrze uwarunkowana. Błędy obu metod (QR i LU) pozostają na niskim poziomie niezależnie od wymiaru $N$.
2. **Przypadek niekorzystny** — macierz Hilberta jest silnie źle uwarunkowana. Wraz ze wzrostem $N$ błędy obu metod rosną dramatycznie, co ilustruje wpływ warunku macierzy na dokładność numeryczną.
3. **QR vs LU** — w przypadku macierzy źle uwarunkowanych faktorizacja QR jest zazwyczaj bardziej stabilna numerycznie niż LU, choć w standardowej implementacji SciPy (z pełnym pivotowaniem) różnice mogą być niewielkie.
4. Błędy zaokrągleń są nieuniknione w arytmetyce zmiennoprzecinkowej; dobór metody i uwarunkowanie danych decydują o tym, czy pozostaną one akceptowalne.
